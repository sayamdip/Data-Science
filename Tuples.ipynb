## Tuple Data Structure
Tuple is an ordered collection of items 

Tuple is immutable


```python
tuple12=("Hello", "World")
tuple12
```

```python
## List Converted To A Tuple
tuple1=tuple([1,2,3,4,5])
tuple1
```

```python
## Tuple Converted To List
lst=list(tuple1)
print(lst)
```

```python
## Accessing Tuple Elements
print(tuple1[:2])
print(tuple1[::2])
print(tuple1[::-1])
```

```python
## Tuple Operations
concat=tuple1+tuple12
print(concat)
```

```python
tuple1*3 ## Get All The Elements 3 Times
```

```python
## Immutable Nature Of Tuples
## Tuples Are Immutable Means We Can't Change The Values

tuple1[0]=10
print(tuple1)
```

```python
## Tuple Methods
print(tuple1)
print(tuple1.count(1))
print(tuple1.index(3)) ## Where This 3 is Present In The Tuple
```

```python
## Packing And Unpacking Tuple

packed_tuple1=1,"Hello",3.14
print(packed_tuple1)
```

```python
## Unpacking A Tuple
a,b,c=packed_tuple1
print(a)
print(b)
print(c)
```

```python
## Unpacking With *
tuple1=(1,2,3,4,5)
first,*middle,last=tuple1
print(first)
print(middle) ## * Does Take From Start Elements Before The Last Element
print(last)
```

```python
## Nested List
list1=[[1,2,3],[4,5,6],("Hello",3.14,"c")]
print(list1[0][1])
print(list1)
print(list1[2][0:])
```

```python
## Nested Tuples
nested_tuple=((1,2,3),("a","b","c"),("True","False"))

## Access The Elements Inside A Tuple
print(nested_tuple[0][1])
print(nested_tuple)
```

```python
## Iterating Over Nested Tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print() ## Helps In Adding A New Line
```
