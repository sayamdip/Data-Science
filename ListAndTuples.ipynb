List Is Mutable Collection Of Data

```python
list=[]
print(type(list))
```

```python
name=["Sayamdip",1,2,3,4]
print(name)
```

```python
## Accessing List elements
fruits=["Apple","Banana","Cherry","Guava"]
for i in fruits:
    print(i)
```

```python
## Indexing Type-1

print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[3])
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
print(fruits[-4])
```

```python
## indexing Type-2

print(fruits[1:])
print(fruits[2:3])
```

```python
## Modify The List Elements
fruits
```

```python
fruits[1]="Watermelon"
print(fruits)
fruits[1:]
```

```python
## List Methods
fruits.append("Pineapple") ## Add An Item To The End
fruits.remove("Pineapple") ## Remove First Occurence Of An Item
fruits.pop() ## Remove Last Item
index=fruits.index("Cherry")
print(index)
```

```python
print(fruits.sort())
fruits.reverse()
fruits
```

```python
## Slicing List
numbers=[1,2,3,4,5,6,7,8]
print(numbers[2:5])
print(numbers[:5])
print(numbers[5:])
print(numbers[::2]) ## Jump Two Steps
print(numbers[::-1]) ## Jumps 1 Step In Reverse
```

```python

## Iterating Over List
for num in numbers:
    print(num)
```

```python
## Iterating With Index
for index,number in enumerate(numbers): ## Enumerate Is The Inbuild Function That Returns The Index And The Value
    print(index,number)
```

```python
## List Comprehension
lst=[]
for i in range(10):
    lst.append(i**2)
print(lst)
```

## List Comprehension
Basic Syntax [expression for item in iterable]

with conditional logic [expression for item in iterable if condition]

Nested List Comprehension [expression for item in iterable for item in iterable]

```python
## Cube Of List Elements
square=[num**3 for num in range(10)]
print(square)
```

```python
## List Comprehension With Condition Even Elements In A List
list=[]
for i in range(10):
    if(i%2==0):
        list.append(i)
print(list)
```

```python
## Nested List Comprehension
list1=[1,2,3,4]
list2=['a','b','c','d']

pair=[(i,j) for i in list1 for j in list2]
print(pair)
```

```python
## List Comprehension With Function Calls
words=["hello","world","python","programming"]
lengths=[len(word) for word in words]
print(lengths)
```
