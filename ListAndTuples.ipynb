# List Data Structure
List Is Mutable Collection Of Data

```python
list=[]
print(type(list))
```

```python
name=["Sayamdip",1,2,3,4]
print(name)
```

```python
## Accessing List elements
fruits=["Apple","Banana","Cherry","Guava"]
for i in fruits:
    print(i)
```

```python
## Indexing Type-1

print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[3])
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
print(fruits[-4])
```

```python
## indexing Type-2

print(fruits[1:])
print(fruits[2:3])
```

```python
## Modify The List Elements
fruits
```

```python
fruits[1]="Watermelon"
print(fruits)
fruits[1:]
```

```python
## List Methods
fruits.append("Pineapple") ## Add An Item To The End
fruits.remove("Pineapple") ## Remove First Occurence Of An Item
fruits.pop() ## Remove Last Item
index=fruits.index("Cherry")
print(index)
```

```python
print(fruits.sort())
fruits.reverse()
fruits
```

```python
## Slicing List
numbers=[1,2,3,4,5,6,7,8]
print(numbers[2:5])
print(numbers[:5])
print(numbers[5:])
print(numbers[::2]) ## Jump Two Steps
print(numbers[::-1]) ## Jumps 1 Step In Reverse
```
