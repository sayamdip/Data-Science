```python
##  Arithmetic operations
##  Simple Calculator
a=int(input("Enter 1st Number= "))
b=int(input("Enter 2nd Number= "))
add=a+b
sub=a-b
mult=a*b
div=a/b
floor_div=a//b
modulo=a%b
expo=a**b

print(add)
print(sub)
print(mult)
print(div)
print(floor_div)
print(modulo)
print(expo)
```
