```python
## For Loop
for i in range (6):
    print(i)
```

```python
for i in range(1,10,2):  ## Here 1 is Initial, 10 Upto And 2 How Much To Jump
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
str="Sayamdip"
for i in str:
    print(i)
```

```python
## While Loop
## It Continues To Execute As Long As The Condition Is True
count=0
while count<5:
    print(count)
    count=count+1
```

```python
count=0
while count%2==0:
    print(count)
    count=count+1
```

```python
## Loop Control Statements

## 1) Break Statement
for i in range(10):
    if i==5:
        break
    print(i)
    i=i+1
```

```python
## Continue
## Continue Statement Skips The Current Iteration And Continues With Next
for i in range(1,10):
    if i%2==0:
        continue
    print(i)
```

```python
## Pass Statment
## The Pass Statment Is A Null Operation; It Does Nothing

for i in range(5):
    if i==3:
        print("The Number Is= ",i)
        pass
    print(i)
```
