```python
## While Loop
## It Executes As Long As The Condition Is True
count=0
while count<5:
    print(count)
    count=count+1
```

```python
count=0
while count%2==0:
    print(count)
    count=count+1
```

```python
## Loop Control Statements

## 1) Break Statement
## In Break Statement If One Condition Becomes True Then It Exits From Its Loop
for i in range(10):
    if i==5:
        break
    print(i)
    i=i+1
```

```python
## Continue
## Continue Statement Skips {The Current Condition Which Becomes True} And {Continues To Print The Rest Which Are Not True}
for i in range(1,10):
    if i%2==0:
        continue
    print(i)
```

```python
## Pass Statment
## The Pass Statment Is A Null Operation; It Does Nothing

for i in range(5):
    if i==3:
        print("The Number Is= ",i)
        pass
    print(i)
```

```python
## Nested Loop
for i in range(1,5):
    for j in range(1,11):
        print(i," X ",j,"= ",i*j)
```

```python
## Calculate The Sum Of N Natural Numbers Using A While And For Loop
## 1) For Loop
a=int(input("Enter A Number= "))
sum=0
for i in range(1,a+1):
    sum=sum+i
print(sum)
```

```python
## 2) While Loop
a=int(input("Enter A Number= "))
sum=0
i=1
while(i<a+1):
    sum=sum+i
    i=i+1
print(sum)
```

```python
## Prime Numbers Between 1 To 100
for i in range(1,101):
    if i>1:
        for j in range(2,i):
            if i%j==0:
                break
        else:  ##       This else belongs to the for j loop, not the if.
               ## 	•	It runs ONLY IF the loop finishes without hitting break
	           ##   •	Meaning: no divisor was found

            print(i)        
```

```python
## All Even Numbers From 1 To 100
for i in range(1,101):
    if(i%2!=0):
        print(i)
```
