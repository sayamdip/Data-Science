# Variables
A variable is a named storage location in memory used to store data that can change during the execution of a program.

```python
# Declaring And Assigning Variables
age=65
height=5.9
name="Sayamdip"
is_student=True

## Printing The Variables
print("Age: ",age)
print("Height: ",height)
print("Name: ",name)
```

```python
## Naming Convention
## Variable Names Should Be Descriptive
## They Must Start With A Letter Or An '_' and contains letter, numbers and underscores
## Variables names case sensitive

# Valid Variable Names
first_name="Sayamdip"
last_name="Dey Chaklader"

# Invalid Variables Names
@name="Sayamdip"
5lastname="Dey Chaklader"
```

```python
# Uderstanding Variables Types
## Python Is Dynamically types, type of a variable is determined at runtime
age=25 #int
height=5.9 #float
name="Sayamdip" #str
is_student=True #bool
```

```python
# Type Checking
print(type(age))
print(type(height))
print(type(name))
print(type(is_student))

#Type Conversion
age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
## Dynamic Typing
## Python Allows The Type Of A Variable To Change As The Program Executes
var=90 #int
print(var,type(var))

var="Hello"
print(var,type(var))

var=3.14
print(var,type(var))
```

```python
##Input
age=int(input("What Is The Age= "))
print(age,type(age))
```

```python
## Simple Calculator
num1=float(input("Enter First Number= "))
num2=float(input("Enter Second Number= "))

sum=num1+num2
diff=num1-num2
mult=num1*num2
div=num1/num2

print("Sum= ",sum)
print("Difference= ",diff)
print("Multiplication= ",mult)
print("Division= ",div)
```
